```python
import numpy as np
```

```python
SPLIT_LEVELS = 3
SPLIT_DEPTH = 2**SPLIT_LEVELS
ARENA_SIZE = 64

arena = np.full(ARENA_SIZE, -1, np.int32)
split = np.full(ARENA_SIZE // SPLIT_DEPTH, -1, np.int8)
END = len(arena)

bump = 0
first = END
current_block = [END] * (SPLIT_LEVELS + 1)
current_in_block = [0] * (SPLIT_LEVELS + 1)

def init():
    global arena, split, bump, first, current_block, current_in_block
    arena[:] = -1
    split[:] = -1
    bump = 0
    first = END
    current_block = [END] * SPLIT_LEVELS
    current_in_block = [0] * SPLIT_LEVELS

def show():
    print(" ".join([y for x in split for y in ["%2d" % x]+ ["  "] * (SPLIT_DEPTH - 1)]))
    print(" ".join("%2d" % x for x in arena))

show()
```

```python
def allocate_block():
    global first, bump
    if first == END:
        if bump == len(split):
            return END
        else:
            out = bump
            bump += 1
    else:
        out = first
        first = arena[first << SPLIT_LEVELS]
    return out
```

```python
def free_block(block):
    global first
    arena[block << SPLIT_LEVELS] = first
    first = block
```
